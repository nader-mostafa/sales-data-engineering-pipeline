```python
## i wanna get discount and sale adn profit 
df["discount"]=df["list_price"]*df["discount_percent"]*0.01

df["sale_price"]=df["list_price"]-df["discount"]


df["profit"]=df["sale_price"]-df["cost_price"]
```

```python
# Load cleaned data into SQL Server

# df.to_sql(
#     "orders",
#     con=engine,
#     if_exists="replace",
#     index=False
# )

from sqlalchemy.types import BigInteger, DateTime, String, Float
```

```python
df.to_sql(
    "orders",
    con=engine,
    if_exists="replace",
    index=False,
    dtype={
        "order_id": BigInteger(),
        "order_date": DateTime(),
        "ship_mode": String(50),
        "segment": String(50),
        "country": String(100),
        "city": String(100),
        "state": String(100),
        "postal_code": BigInteger(),
        "region": String(50),
        "category": String(50),
        "sub_category": String(50),
        "product_id": String(50),
        "quantity": BigInteger(),
        "discount": Float(),
        "sale_price": Float(),
        "profit": Float()
    }
)
```
